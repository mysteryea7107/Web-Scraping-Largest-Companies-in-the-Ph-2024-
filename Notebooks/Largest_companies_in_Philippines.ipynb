```python
from bs4 import BeautifulSoup
import requests
```

```python
url = 'https://en.wikipedia.org/wiki/List_of_largest_companies_in_the_Philippines'

headers = {
    "User-Agent": (
        "MyDataProject/1.0 (https://example.com/contact; your-email@example.com)"
    )
}

response = requests.get(url, headers=headers)
soup = BeautifulSoup(response.text, "html.parser")
```

```python
print(soup.prettify())

```

```python
titles = [title.text.strip() for title in table.find_all('th')]
print(titles)
```

```python
import pandas as pd
```

```python
pd.DataFrame(columns=titles)
```

```python
tables = soup.find_all("table", {"class": "wikitable"})
```

```python
target_table = None
```

```python
# Loop through the tables to find the correct one using a unique header keyword
for table in tables:
  # 'Employees' only exists in the first table's headers
  if "Employees" in table.text:
    target_table = table
    break
```

```python
# Extract the rows safely from the target table
if target_table:
  rows = target_table.find_all("tr")
  for row in rows:
    cols = row.find_all("td")
    # Only process rows with <td> cells (skips the header <th> row)
    if cols:
      individual_row_data = [data.text.strip() for data in cols]
      print(individual_row_data)
else:
  print("Target table not found.")
```
